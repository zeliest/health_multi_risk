```python
from pathlib import Path
import sys
sys.path.insert(0, '..')
sys.path.insert(0, '../exposure_to_hazards/')
from config import FIGURES, DATA_DIR 

import itertools
from climada.engine import Impact
combinations = []

hazards = ['RF','EP','DR','HW','FI','TC']

for r in range(1, len(hazards) + 1):
    combinations.extend(list(itertools.combinations(hazards, r)))

combinations = ["-".join(combi) for combi in combinations]

IMPACT_DIR = DATA_DIR / "impacts"

imp_combined = {}
for age in ['all']:
    imp_combined[age] = {}
    for combi in combinations:
        imp_combined[age][combi] = Impact.from_csv(IMPACT_DIR / f"csv/{combi}_{age}_2003_2022.csv")
        imp_combined[age][combi].imp_mat = Impact.read_sparse_csr(IMPACT_DIR / f"npz/{combi}_{age}_2003_2022.npz")
        imp_combined[age][combi]

```

```python
from pathlib import Path
combinations_ec = []
hazards = ['RF_ec','EP_ec','DR_ec','HW_ec','FI_ec','TC_ec']

for r in range(1, len(hazards) + 1):
    combinations_ec.extend(list(itertools.combinations(hazards, r)))

combinations_ec = ["-".join(combi) for combi in combinations_ec]

imp_combined_ec = {}
for age in ['all']:
    imp_combined_ec[age] = {}
    for combi in combinations_ec:
        imp_combined_ec[age][combi] = Impact.from_csv(IMPACT_DIR / f"csv/{combi}_{age}_2003_2022.csv")
        imp_combined_ec[age][combi].imp_mat = Impact.read_sparse_csr(IMPACT_DIR / f"npz/{combi}_{age}_2003_2022.npz")
        imp_combined_ec[age][combi]
```

```python
# Define the criteria for different hazard combinations
hazard_lengths = {
    "single": 2,
    "two_or_more": 5,  # Assuming "two or more" refers to a tuple length of 5
    "three_or_more": 8
}

# Helper function to get tuple keys based on hazard lengths
def get_tuple_keys(imp_dict, age, length):
    return [key for key in imp_dict[age].keys() if len(key.replace("_ec", "")) == length]

# Dictionary to store the keys and impacts
imp_combined_dicts = {
    "single": {},
    "two_or_more": {},
    "three_or_more": {}
}

# Populate the dictionaries with the respective hazard combinations
for key, length in hazard_lengths.items():
    imp_combined_dicts[key] = {
        age: {
            k: imp_combined[age][k]
            for k in get_tuple_keys(imp_combined, age, length)
            if imp_combined[age][k].imp_mat.nnz > 0
        }
        for age in imp_combined
    }

    imp_combined_dicts[key + "_ec"] = {
        age: {
            k: imp_combined_ec[age][k]
            for k in get_tuple_keys(imp_combined_ec, age, length)
            if imp_combined_ec[age][k].imp_mat.nnz > 0
        }
        for age in imp_combined_ec
    }

# Combine the impacts for "Three or more" and "One or more" (and optionally "Two or more")
from multi_risk_methods import combine_impacts

for age in imp_combined:
    imp_combined[age]['Three or more'] = combine_impacts(imp_combined_dicts["three_or_more"][age], how="max")
    imp_combined_ec[age]['Three or more'] = combine_impacts(imp_combined_dicts["three_or_more_ec"][age], how="max")
    
    imp_combined[age]['One or more'] = combine_impacts(imp_combined_dicts["single"][age], how="max")
    imp_combined_ec[age]['One or more'] = combine_impacts(imp_combined_dicts["single_ec"][age], how="max")
    
    imp_combined[age]['Two or more'] = combine_impacts(imp_combined_dicts["two_or_more"][age], how="max")
    imp_combined_ec[age]['Two or more'] = combine_impacts(imp_combined_dicts["two_or_more_ec"][age], how="max")
```

```python
import pandas as pd
import datetime 
dfs = []

# Define the special combinations
special_combinations = ['Three or more', 'One or more', 'Two or more']

for combi in imp_combined[age]:
    
    # Handle the '_ec' suffix for combinations
    if combi in special_combinations:
        combi_ec = combi  # No need to modify special combinations
    else:
        combi_ec = "-".join(["".join([c, "_ec"]) for c in combi.split("-")])
    
    # Create the DataFrame for each combination
    df = pd.DataFrame({
        'date': imp_combined[age][combi].date.astype(int),
        'at_event_population_growth': imp_combined[age][combi].at_event,
        'at_event_constant_population': imp_combined_ec[age][combi_ec].at_event,
        'combination': combi,
        'age_group': age
    })
    dfs.append(df)

# Concatenate all the individual DataFrames into one
df_concat = pd.concat(dfs)

# Convert ordinal date to datetime
df_concat['date'] = [datetime.datetime.fromordinal(int(date)) for date in df_concat.date]

# Save the concatenated DataFrame to a CSV file
df_concat.to_csv(IMPACT_DIR / "monthly_exposure_global_all_combinations_2003_2021.csv")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
import numpy as np
from scipy.stats import linregress

# Paths and font size setup
plt.rcParams.update({'font.size': 18})  # Increased font size

df_concat = pd.read_csv(IMPACT_DIR / "monthly_exposure_global_all_combinations_2003_2021.csv", index_col=0)
df_concat['date'] = pd.to_datetime(df_concat['date'])

combinations_to_plot = ['HW', 'FI', 'HW-FI', 'DR-HW-FI']  # Adjust as necessary to select the combinations you want

fig, axs = plt.subplots(2, 2, figsize=(20, 10))  # Create a 2x2 grid of subplots
axs = axs.flatten()  # Flatten the 2x2 array to a 1D array for easy iteration

# Color setup for bars and lines
color_map = {
    'HW': 'blue', 
    'FI': 'green', 
    'HW-FI': 'red', 
    'DR-HW-FI': 'purple'
}

for ax, combi in zip(axs, combinations_to_plot):
    df_combi = df_concat[df_concat.combination == combi]
    if df_combi["at_event_population_growth"].sum() < 10000:  # Skip combination that are not affecting many people
        ax.set_visible(False)  # Hide the subplot if the combination is not affecting many people
        continue

    # Convert date for linear regression
    date_nums = df_combi['date'].map(pd.Timestamp.toordinal)

    # Linear regression calculations for both datasets
    slope_growth, intercept_growth, r_value_growth, p_value_growth, std_err_growth = linregress(date_nums, df_combi['at_event_population_growth'])
    slope_constant, intercept_constant, r_value_constant, p_value_constant, std_err_constant = linregress(date_nums, df_combi['at_event_constant_population'])

    # Print the results
    print(f"Combination: {combi}")
    print(f"Slope for growth: {slope_growth}, P-value for growth: {p_value_growth}")
    print(f"Slope for constant: {slope_constant}, P-value for constant: {p_value_constant}")

    # Plotting bars
    bar_width = 15
    ax.bar(df_combi.date, df_combi.at_event_population_growth/1e6, width=bar_width, alpha=0.8, label="With population growth", color=colors_dict_combi[combi])
    ax.bar(df_combi.date, df_combi.at_event_constant_population/1e6, width=bar_width, alpha=0.8, label="Constant population (2003 levels)", color='lightgrey')

    # Plotting linear fit lines
    ax.plot(df_combi.date, (intercept_growth + slope_growth * date_nums)/1e6, '--', color='darkgrey', label="Trend with population growth")
    ax.plot(df_combi.date, (intercept_constant + slope_constant * date_nums)/1e6, '--', color='black', label="Trend with constant population")

    ax.set_title(combi)
    ax.set_xlabel("Date")
    ax.set_ylabel("Population exposed (millions)")
    ax.legend()

plt.tight_layout()
#plt.savefig(FIGURES / "trends.png", dpi=300)
plt.show()
```
