```python
import sys
import numpy as np
from datetime import datetime, timedelta
from netCDF4 import Dataset, date2num
sys.path.insert(0, '../..')
from config import DATA_DIR

IMP_DIR = DATA_DIR / "impacts"

# Loop over the ages and hazard
for age in imp_combined:
    for haz in ['RF','EP','DR','HW','FI','TC']:
        print(f"Processing hazard: {haz} for age group: {age}")
        
        # Extract sparse matrix and coordinates
        sparse_matrix = imp_combined[age][haz].imp_mat
        coordinates = imp_combined[age][haz].coord_exp
        dates = imp_combined[age][haz].date
        
        # Extract unique latitudes and longitudes
        latitudes = np.unique(coordinates[:, 0])
        longitudes = np.unique(coordinates[:, 1])
        
        # Initialize a dense 3D array (time x latitude x longitude)
        time_dim_size = sparse_matrix.shape[0]
        dense_data = np.full((time_dim_size, len(latitudes), len(longitudes)), np.nan)

        # Populate the dense array
        for i, j, value in zip(*sparse_matrix.nonzero(), sparse_matrix.data):
            if i < time_dim_size:  # Safety check for time dimension boundary
                lat_idx = np.where(latitudes == coordinates[j, 0])[0][0]
                lon_idx = np.where(longitudes == coordinates[j, 1])[0][0]
                dense_data[i, lat_idx, lon_idx] = value
            else:
                print(f"Skipping out-of-bounds index: i={i}, j={j}")

        # Replace any remaining NaNs with a fill value
        dense_data = np.nan_to_num(dense_data, nan=-9999.0)
        
        # Convert ordinal dates to datetime
        datetime_dates = [datetime(1, 1, 1) + timedelta(days=int(d)) for d in dates]
        
        # Create NetCDF file
        output_file = IMP_DIR / f'netcdf/{age}_exposed_{haz}_2003_2021.nc'
        with Dataset(output_file, 'w', format='NETCDF4') as ncfile:
            print(f"Creating NetCDF file: {output_file}")

            # Define dimensions
            time_dim = ncfile.createDimension('time', dense_data.shape[0])
            lat_dim = ncfile.createDimension('latitude', len(latitudes))
            lon_dim = ncfile.createDimension('longitude', len(longitudes))
        
            # Define variables
            time_var = ncfile.createVariable('time', 'f8', ('time',))
            lat_var = ncfile.createVariable('latitude', 'f4', ('latitude',))
            lon_var = ncfile.createVariable('longitude', 'f4', ('longitude',))
            data_var = ncfile.createVariable('data', 'f8', ('time', 'latitude', 'longitude'), fill_value=-9999.0)
        
            # Add metadata
            ncfile.description = "Sparse data converted to standard NetCDF4 format"
            ncfile.history = "Created on " + datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        
            # Populate variables
            time_var.units = "days since 1970-01-01 00:00:00"
            time_var.calendar = "proleptic_gregorian"
            time_var[:] = date2num(datetime_dates, units=time_var.units, calendar=time_var.calendar)
        
            lat_var[:] = latitudes
            lat_var.units = "degrees_north"
        
            lon_var[:] = longitudes
            lon_var.units = "degrees_east"
        
            # Assign data after final checks
            print("Final check before writing data:")
            print(f"Shape of dense_data: {dense_data.shape}")
            print(f"Number of -9999.0 (fill values): {np.sum(dense_data == -9999.0)}")
            
            data_var[:, :, :] = dense_data
            data_var.units = "People Exposed"

        print(f"NetCDF file '{output_file}' created successfully!")
```
